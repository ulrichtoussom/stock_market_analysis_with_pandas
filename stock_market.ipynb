```python
# Remplacez les valeurs manquantes de `prix_unitaire` par le prix moyen **du même ticker**

transactions['prix_unitaire'] = transactions.groupby('ticker')['prix_unitaire'].transform(lambda x : x.fillna(x.mean()))
```

```python
transactions['prix_unitaire'].isna().sum()
```

1.  Créez une colonne `montant_brut` = `quantite` × `prix_unitaire` ;
2. Créez une colonne `montant_net` qui vaut `montant_brut + frais` pour un `'Achat'` (les frais s'ajoutent à ce qu'on paie), et `montant_brut - frais` pour une `'Vente'` (les frais se déduisent de ce qu'on reçoit). Indice : utilisez `np.where(transactions['type_operation'] == 'Achat', ..., ...)`.

```python
transactions['montant_brut'] = transactions['quantite'] * transactions['prix_unitaire']
transactions['montant_net'] = np.where(
    transactions['type_operation']=='Achat',
    transactions['montant_brut'] + transactions['frais'],
    transactions['montant_brut'] - transactions['frais']

)

transaction_new = transactions[['montant_brut', 'montant_net']]
display(transactions.head(5))
```
